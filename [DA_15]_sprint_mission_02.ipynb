## 문제 1. 3의 배수 출력하기

### 🧩 문제 설명
1부터 n까지의 숫자 중에서 3의 배수만 출력하는 함수를 작성하세요.

### 👉🏻 함수 설명
`print_multiples_of_three(n)`
- `n` (int) : 3의 배수를 구할 범위의 최댓값

### 💡 입출력 예시  

***예시1**

입력: `7`

출력
>```
>3
>6
>```


***예시2**

입력: `30`

출력
>```
>3
>6
>9
>12
>15
>18
>21
>24
>27
>30
>```

````python
def print_multiples_of_three(n):
    # 여기에 함수의 내용을 작성하세요
    for i in range(1, n+1):
        if i % 3 == 0:
            print(i)
````

````python
# 이 코드를 실행해서 함수를 확인합니다
print_multiples_of_three(7)
````

## 문제 3. 모음 제거하기

### 🧩 문제 설명
문자열에서 모음(a, e, i, o, u)을 제거한 문자열을 반환하는 함수를 작성하세요.

### 👉🏻 함수 설명
`remove_vowels(s)`
- `s` (string) : 영어 소문자 알파벳으로 이루어진 문자열

### 💡 입출력 예시

***예시1**
- 입력: `"codeit"`
  
- 출력: `"cdt"`


***예시2**
- 입력: `"python programming"`
  
- 출력: `"pythn prgrmmng"`

````python
def remove_vowels(s):
    # 여기에 함수의 내용을 작성하세요
    vowels = "aeiou"

    result = ""
    for char in s:
        if char not in vowels:
            result += char

    return result
````

````python
# 이 코드를 실행해서 함수를 확인합니다
remove_vowels("codeit")
````

## 문제 4. 숫자 비교하기

### 🧩 문제 설명

두 개의 숫자를 [큰 숫자, 작은 숫자] 순서대로 반환하는 함수를 작성하세요.  
단, 동일한 숫자를 입력한 경우에는 동일한 숫자 2개를 그대로 반환합니다.

### 👉🏻 함수 설명

`compare_numbers(num1, num2)`
- `num1` (int) : 첫 번째 숫자
- `num2` (int) : 두 번째 숫자

### 💡 입출력 예시

***예시1**
- 입력: `20, 30`

- 출력: `30, 20`

***예시2**
- 입력: `10, 10`

- 출력: `10, 10`


````python
def compare_numbers(num1, num2):
    # 여기에 함수의 내용을 작성하세요
    if num1 >= num2:
        return num1, num2
    else:
        return num2, num1
````

````python
# 이 코드를 실행해서 함수를 확인합니다
print(compare_numbers(20, 30))
print(compare_numbers(10, 10))
````

## 문제 6. 총 합계 구하기

### 🧩 문제 설명

0부터 사용자가 입력한 숫자 `n`까지의 합을 구하는 함수를 작성하세요.

### 👉🏻 함수 설명

`sum_up_to_n(n)`
- `n` (int) : 1 이상의 정수

### 💡 입출력 예시

***예시1**

- 입력: `3`
  
- 출력: `6`

***예시2**

- 입력: `10`
  
- 출력: `55`


````python
def sum_up_to_n(n):
    # 여기에 함수의 내용을 작성하세요
    total = 0

    for i in range(1, n + 1):
        total += i

    return total
````

## 문제 7. 짝수 합 구하기

### 🧩 문제 설명

숫자 `n`까지의 짝수 합을 구하는 함수를 작성하세요.

### 👉🏻 함수 설명

`sum_even_numbers(n)`
- `n` (int) : 1 이상의 정수

### 💡 입출력 예시

***예시1**

- 입력: `9`
  
- 출력: `"1부터 9까지의 짝수 합은 20입니다."`

***예시2**

- 입력: `15`
  
- 출력: `"1부터 15까지의 짝수 합은 56입니다."`


````python
def sum_even_numbers(n):
    # 여기에 함수의 내용을 작성하세요
    total = 0

    for i in range(1, n + 1):
        if i % 2 == 0:
            total += i  # total = total + i

    return f"1부터 {n}까지의 짝수 합은 {total}입니다."
````

````python
# 이 코드를 실행해서 함수를 확인합니다
sum_even_numbers(9)
````

### 👉🏻 함수 설명

`divide_numbers(num1, num2)`
- `num1` (int) : 첫 번째 숫자
- `num2` (int) : 두 번째 숫자
- `num1`을 `num2`로 나누는 함수입니다.
- 예외가 발생하면 `"잘못된 입력입니다."`를 반환합니다.
- `try-except` 구문을 활용하는 것도 좋습니다. 잘 모르겠다면 검색을 시도해 보세요.

### 💡 입출력 예시

***예시1**

- 입력: `42, 7`

- 출력: `6`

***예시2**

````python
42 / 7
````

````python
135 / 0
````

````python
def divide_numbers(num1, num2):
    # 여기에 함수의 내용을 작성하세요
    try:
        return num1 / num2
    except ZeroDivisionError:
        return "잘못된 입력입니다."
````

## 문제 9. BMI 계산하기

### 🧩 문제 설명

사용자가 입력한 키와 몸무게 값을 이용해 BMI를 계산하고 비만도를 판별하는 함수를 작성하세요.

### 👉🏻 함수 설명
`check_obesity(height, weight)`
- `height` (int): 키 정보 (cm)
- `weight` (int): 몸무게 정보 (kg)
- 다음 공식을 사용하여 BMI를 계산합니다.
$$ BMI = \frac{\text{체중 (kg)}}{\text{신장 (m)}^2} $$


- BMI 값을 기준으로 비만도를 다음과 같이 분류하여 반환합니다.
    - 18.5 미만 → '저체중'
    - 18.5 이상 23 미만 → '정상'
    - 23 이상 25 미만 → '과체중'
    - 25 이상 → '비만'

### 💡 입출력 예시

***예시1**

* 입력: `175, 60`
* 출력: `정상`

***예시2**

* 입력: `180, 88`
* 출력: `비만`


````python
def check_obesity(height, weight):
    # 여기에 함수의 내용을 작성하세요
    # 신장(cm -> m 환산)
    height_m = height/100

    # BMI 계산
    bmi = weight / (height_m ** 2)

    # 비만도 분류
    if bmi < 18.5:
        return "저체중"
    elif bmi < 23:
        return "정상"
    elif bmi < 25:
        return "과체중"
    else:
        return "비만"
````

````python
# 이 코드를 실행해서 함수를 확인합니다
print(check_obesity(175, 60))
print(check_obesity(180, 88))
````

## 문제 10. 문자열 압축하기

### 🧩 문제 설명

주어진 문자열에서 연속해서 같은 문자가 반복되는 경우, 그 문자와 반복 횟수를 이용해 문자열을 압축하는 함수를 작성하세요.
예를 들어 "aabcccccaaa"는 "a2b1c5a3`로 압축될 수 있습니다.
단, 압축 결과의 길이가 원본 문자열보다 짧을 경우에만 압축 결과를 반환하고, 그렇지 않으면 원래 문자열을 반환합니다.

### 👉🏻 함수 설명

`compress_string(s)`
- `s` (string) : 압축하고자 하는 문자열

### 💡 입출력 예시

***예시1**
- 입력: `"aaabbaaa"`
  
- 출력: `"a3b2a3"`

***예시2**
- 입력: `"aabcccccaaa"`
  
- 출력: `"a2b1c5a3"`


````python
def compress_string(s):
    # 여기에 함수의 내용을 작성하세요
    count = 1
    compressed_string = ""

    for i in range(len(s)):
        # (1) 현재 문자와 다음 문자가 같으면 : 1부터 시작해서 1씩 증가
        if (i+1 < len(s)) and (s[i] == s[i+1]):
            count += 1
        # (2) 현재 문자와 다음 문자가 다르면 : 문자+개수 출력하고, 개수 1로 초기화
        else:
            compressed_string += s[i] + str(count)
            count = 1

    # 원본 문자열과 압축된 문자열의 길이 비교
    if len(compressed_string) < len(s):
        return compressed_string
    else:
        return s
````

````python
# 이 코드를 실행해서 함수를 확인합니다
compress_string("aaabbaaa")
````

```
1번째 숫자를 입력하세요: 2
2번째 숫자를 입력하세요: 8
3번째 숫자를 입력하세요: 0
3S 0B

축하합니다. 3번 만에 숫자 3개의 값과 위치를 모두 맞히셨습니다.
```

````python
from random import randint


def generate_numbers():
    # 여기에 함수의 내용을 작성하세요
    numbers = []

    while len(numbers) < 3:
        new_number = randint(0, 9)
        if new_number not in numbers:
            numbers.append(new_number)

    return numbers


def take_guess():
    # 여기에 함수의 내용을 작성하세요
    new_guess = []
    while len(new_guess) < 3:
        num = int(input(f"{len(new_guess) + 1}번째 수를 입력하세요: "))

        if num < 0 or num > 9:
            print("0에서 9까지의 수를 입력해 주세요!")
        elif num in new_guess:
            print("중복되는 숫자입니다. 다시 입력하세요.")
        else:
            new_guess.append(num)

    return new_guess


def get_score(guesses, solution):
    # 여기에 함수의 내용을 작성하세요
    strike_count = 0
    ball_count = 0

    for i in range(3):
        if guesses[i] == solution[i]:
            strike_count += 1
        elif guesses[i] in solution:
            ball_count += 1

    return strike_count, ball_count
````

````python
# 세 함수를 완성한 후, 아래 코드를 실행해서 숫자 야구 게임이 제대로 작동하는지 확인해 보세요.
answer = generate_numbers()
answer
````

````python
tries = 0

while True:
    user_guess = take_guess()
    s, b = get_score(user_guess, answer)

    print(f"{s}S {b}B\n")
    tries += 1

    if s == 3:
        break

print(f"축하합니다. {tries}번 만에 세 숫자의 값과 위치를 모두 맞히셨습니다.")
````
